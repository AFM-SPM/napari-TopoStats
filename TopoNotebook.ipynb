```python
import napari
from skimage.io import imread
import numpy
import copy
```

```python
viewer = napari.Viewer()
```

```python
original = imread("H:\\NapariWorkshop\\firezproject.tif")
viewer.add_image(original)
```

```python
numslices = 256
```

```python
shape = original.shape
print(shape)
stacked = numpy.empty((numslices, shape[0], shape[1]))
print(stacked.shape)
```

```python
for z in range(0, numslices):
	dup = copy.deepcopy(original)
	dup[dup > z] = z
	stacked[z,:,:] = dup
```

```python
viewer.add_image(stacked, name='stacked_inferno', colormap='inferno')
```
