```python
import napari
from skimage.io import imread
import numpy
import copy
```

```python
viewer = napari.Viewer()
```

```python
mesh = imread("H:\\NapariWorkshop\\mesh_3D_stack.tif")
viewer.add_image(mesh)
```

```python
shape = mesh.shape
quickmesh = copy.deepcopy(mesh)
```

```python
print(shape)
```

```python
for z in range(0, shape[0]):
	quickmesh[z,:,:] = (mesh[z,:,:]/255) * z
```

```python
viewer.add_image(mesh)
viewer.add_image(quickmesh, name='quickmesh', colormap='inferno')
```
